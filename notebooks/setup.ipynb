# Check the QAChat setup

This example chooses MLX or PyTorch and calculates `1 + 2 + 3`. It downloads no models.

Run the setup script first. Select QAChat's `.venv` as this notebook's kernel, then run one cell at a time with Shift+Enter.

```python
from pathlib import Path
import platform
import sys

# Find the repository whether Jupyter starts in QAChat or its notebooks folder.
root = Path.cwd()
if not (root / "scripts/check_setup.py").is_file():
    root = root.parent
assert (root / "scripts/check_setup.py").is_file(), "Open this notebook from QAChat."
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from scripts.check_setup import choose_backend, check_setup

print("Python:", sys.executable)
print("Computer:", platform.system(), platform.machine())
assert Path(sys.prefix).resolve() == (root / ".venv").resolve(), "Select QAChat's .venv kernel."
```

## 1. Choose a backend

Keep `auto` for the first run. On an Apple Silicon Mac it chooses MLX; on supported Windows/Linux machines it chooses PyTorch. You can change it to `torch` to try the other installed path on your Mac.

This cell only makes a choice. It does not import either model library yet.

```python
BACKEND = "auto"
selected = choose_backend(platform.system(), platform.machine(), BACKEND)
print("Chosen backend:", selected)
```

## 2. Run the small check

The same function is used by the setup script, command-line check, and debugger. Expect `result` to be `6.0`. `metal`, `mps`, and `cuda` mean accelerator paths; `cpu` means the ordinary processor.

For line-by-line debugging, open `scripts/check_setup.py` and follow the breakpoint instructions in `DEVELOPER_NOTES.md`.

```python
report = check_setup(BACKEND)
assert report["result"] == 6.0
report
```

## 3. Understand other platform choices

These are simulated inputs to the selection function, not tests on real Windows/Linux computers. No software is installed by this cell.

```python
for system, machine in [("Darwin", "arm64"), ("Windows", "AMD64"), ("Linux", "x86_64")]:
    print(system, machine, "->", choose_backend(system, machine))
```
