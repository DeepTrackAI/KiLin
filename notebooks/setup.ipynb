# Check the QAChat setup

This example chooses MLX or PyTorch and adds two numbers you choose. It downloads no models.

Run the setup script first. Select QAChat's `.venv` as this notebook's kernel, then run one cell at a time with Shift+Enter.

```python
from pathlib import Path
import math
import platform
import sys

# Find the repository whether Jupyter starts in QAChat or its notebooks folder.
root = Path.cwd()
if not (root / "scripts/check_setup.py").is_file():
    root = root.parent
assert (root / "scripts/check_setup.py").is_file(), "Open this notebook from QAChat."
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from scripts.check_setup import choose_backend, check_setup

print("Python:", sys.executable)
print("Computer:", platform.system(), platform.machine())
assert Path(sys.prefix).resolve() == (root / ".venv").resolve(), "Select QAChat's .venv kernel."
```

## 1. Choose a backend

Keep `auto` for the first run. On an Apple Silicon Mac it chooses MLX; on supported Windows/Linux machines it chooses PyTorch. You can change it to `torch` to try the other installed path on your Mac.

This cell only makes a choice. It does not import either model library yet.

```python
BACKEND = "auto"
selected = choose_backend(platform.system(), platform.machine(), BACKEND)
print("Chosen backend:", selected)
```

## 2. Enter two numbers

Edit `A` and `B` below. Start with `2` and `3`, then try `8` and `4`. After changing them, run this cell and the check cell below again.

```python
A = 2
B = 3
```

## 3. Check the answer

`check_setup` asks the selected backend to add your numbers. `expected = A + B` separately calculates the answer with ordinary Python. The assertion checks that they agree: `2 + 3` should give `5`, and `8 + 4` should give `12`.

`math.isclose` allows tiny rounding differences for decimal numbers, such as `0.1 + 0.2`. A wrong answer stops the cell with an error.

The same function is used by the setup script, command-line check, and debugger. In the report, `metal`, `mps`, and `cuda` mean accelerator paths; `cpu` means the ordinary processor. For debugging, open `scripts/check_setup.py` and find the comments marked `Breakpoint`.

```python
report = check_setup(BACKEND, a=A, b=B)
expected = A + B  # Python's answer, separate from the backend's answer.
assert math.isclose(report["result"], expected, rel_tol=1e-6, abs_tol=1e-6)
print(f"Checked: {A} + {B} = {report['result']:g}")
report
```

## 4. Understand other platform choices

These are simulated inputs to the selection function, not tests on real Windows/Linux computers. No software is installed by this cell.

```python
for system, machine in [("Darwin", "arm64"), ("Windows", "AMD64"), ("Linux", "x86_64")]:
    print(system, machine, "->", choose_backend(system, machine))
```
