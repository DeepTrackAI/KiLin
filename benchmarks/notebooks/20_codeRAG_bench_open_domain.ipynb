# CodeRAG-Bench evaluation: open-domain (ODEX)

[CodeRAG-Bench](https://github.com/code-rag-bench/code-rag-bench) is a published benchmark for retrieval-augmented code generation; this notebook runs its open-domain task against QAChat's own real retrieval pipeline.

**The Problem.** Given a natural-language intent ("decode a hex string `'4a4b4c'` to UTF-8"), can QAChat's real retrieval pipeline find the library documentation snippet that actually answers it, out of a corpus where nothing in the query is copied from the correct document?

**The Data.** ODEX, CodeRAG-Bench's open-domain task: 201 queries with an annotated relevant document (out of 439; the rest have no annotation and are dropped, matching CodeRAG-Bench's own practice), searched against 34,003 Python library documentation snippets from [`neulab/docprompting-conala`](https://huggingface.co/datasets/neulab/docprompting-conala).

**The Metrics.** NDCG, MRR, Recall, and Precision at several cutoffs, computed via `pytrec_eval` through `evaluate_retrieval`.

**What's different from the original benchmark.** The queries, corpus, and relevance judgments are unmodified from CodeRAG-Bench's own data. The retrieval is QAChat's real pipeline: real chunking (`chunk_source_units`) instead of treating each document as one indivisible unit, and hybrid fusion (semantic + lexical through `hybrid_search`) instead of a single retriever -- no code channel, since ODEX's corpus is prose documentation, not parseable code, so there's nothing for `code_search` to resolve. Because QAChat's chunker can split one corpus document into several chunks, the original document-level relevance judgments no longer line up directly with what's in the index, so each document's relevance is propagated to every chunk built from it before scoring -- a hit on any fragment of the right document still counts as finding it. This isn't a neutral fix, though, so the notebook also runs a whole-document baseline that matches CodeRAG-Bench's own un-chunked corpus exactly, to give a number that's directly comparable to the published range.

## 3. Build the ODEX task

Builds queries, a 34,003-document corpus, and CodeRAG-Bench's own qrels annotation from three separate sources into one `BenchmarkTask`. Only 201 of 439 ODEX examples have an annotated relevant document; the rest are dropped.

```python
import json
import urllib.request

ODEX_DOCS_URL = (
    "https://raw.githubusercontent.com/code-rag-bench/code-rag-bench/"
    "main/retrieval/datasets/canonical/odex_docs.json"
)

with urllib.request.urlopen(ODEX_DOCS_URL) as response:
    odex_docs_text = response.read().decode("utf-8")

odex_docs_rows = [
    json.loads(line) for line in odex_docs_text.splitlines() if line.strip()
]
print(f"ODEX canonical rows: {len(odex_docs_rows)}")
```

```python
# neulab/docprompting-conala ships a legacy loading script, which
# modern `datasets` refuses to execute ("Dataset scripts are no
# longer supported"). Hugging Face auto-converts every dataset to
# Parquet regardless, so this loads that conversion directly instead
# of going through the repository's own script.
DOCPROMPTING_DOCS_PARQUET_URL = (
    "https://huggingface.co/datasets/neulab/docprompting-conala/"
    "resolve/refs%2Fconvert%2Fparquet/docs/train/0000.parquet"
)

doc_rows = datasets.load_dataset(
    "parquet", data_files=DOCPROMPTING_DOCS_PARQUET_URL
)["train"]

doc_index_by_id = {
    doc_id: index for index, doc_id in enumerate(doc_rows["doc_id"])
}

odex_corpus: dict[str, dict[str, str]] = {
    str(index): {"title": doc_id, "text": doc_content}
    for index, (doc_id, doc_content) in enumerate(
        zip(doc_rows["doc_id"], doc_rows["doc_content"])
    )
}
print(f"Documentation corpus: {len(odex_corpus)} documents")
```

```python
# Same Parquet-conversion workaround as the corpus load above --
# neulab/odex also ships a loading script.
ODEX_EN_TEST_PARQUET_URL = (
    "https://huggingface.co/datasets/neulab/odex/"
    "resolve/refs%2Fconvert%2Fparquet/en/test/0000.parquet"
)

odex_rows = datasets.load_dataset(
    "parquet", data_files=ODEX_EN_TEST_PARQUET_URL
)["train"]
assert len(odex_rows) == len(odex_docs_rows), (
    "neulab/odex[en][test] and the canonical annotation file must be in "
    "the same order -- row counts differ, something upstream changed."
)

odex_queries: dict[str, str] = {}
odex_qrels: dict[str, dict[str, int]] = {}
missing_titles = 0

for index, (row, annotated) in enumerate(zip(odex_rows, odex_docs_rows)):
    matched = {
        str(doc_index_by_id[doc["title"]]): 1
        for doc in annotated["docs"]
        if doc["title"] in doc_index_by_id
    }
    missing_titles += len(annotated["docs"]) - len(matched)
    if not matched:
        # No annotated doc, or none of its titles matched the corpus --
        # either way this query cannot be scored, so it is dropped rather
        # than kept with an empty qrels entry.
        continue

    query_id = f"{index}_{row['task_id']}"
    odex_queries[query_id] = row["intent"]
    odex_qrels[query_id] = matched

print(
    f"ODEX queries with a known relevant document: {len(odex_queries)} "
    f"of {len(odex_rows)}"
)
if missing_titles:
    print(
        f"Annotated doc titles not found in the corpus: {missing_titles} "
        "(dropped from qrels; check the corpus load above if this looks "
        "large)"
    )
```

```python
from benchmarks.common import BenchmarkTask

odex_task = BenchmarkTask(
    "odex", odex_corpus, odex_queries, odex_qrels
)
print(
    f"{odex_task.name}: {len(odex_task.queries)} queries, "
    f"{len(odex_task.corpus)} corpus documents"
)
```

## 5. Turn the corpus into a real QAChat project

QAChat's `.txt` extension splits each file into paragraph units at blank lines, so one document can produce more than one chunk -- including the handful of documents running up to ~386,000 characters (median ~318), which real chunking simply splits into more, normally-sized chunks.

```python
from qachat.chunking import Chunk, chunk_source_units
from qachat.repository import scan_repository
from qachat.sources import extract_source_units, read_repository_files

# scan_repository defaults to max_entries=10_000 as a safety guard against
# accidentally pointing it at a huge or wrong folder. ODEX's repository has
# one file per corpus document, so we raise the limit deliberately to match
# the corpus we just wrote, rather than picking an arbitrary large number.
repository_report = scan_repository(
    project.source,
    extensions=profile.extensions,
    max_file_bytes=profile.max_file_bytes,
    max_entries=len(path_to_doc_id) + 100,
)
source_files = read_repository_files(repository_report)
source_units = extract_source_units(source_files)
odex_chunks = chunk_source_units(
    source_units,
    count_tokens=embedding_session.count_tokens,
    target_tokens=profile.target_tokens,
    max_tokens=profile.max_tokens,
    overlap_tokens=profile.overlap_tokens,
)

print(
    f"{len(source_files)} files -> {len(source_units)} source units -> "
    f"{len(odex_chunks)} chunks"
)
```

## 6. Propagate document-level qrels to chunk level

Since real chunking can split one document into several chunks, give every resulting chunk its source document's relevance -- so a hit on any fragment of the right document still counts as finding it.

```python
chunks_by_doc_id: dict[str, list[Chunk]] = {}
for chunk in odex_chunks:
    doc_id = path_to_doc_id.get(chunk.path.name)
    if doc_id is None:
        continue
    chunks_by_doc_id.setdefault(doc_id, []).append(chunk)

doc_ids = list(odex_task.corpus)
counts = [len(chunks_by_doc_id.get(doc_id, [])) for doc_id in doc_ids]
split = sum(1 for count in counts if count > 1)
empty = sum(1 for count in counts if count == 0)
print(
    f"odex: {len(doc_ids)} documents, {split} split into >1 chunk, "
    f"{empty} produced 0 chunks (max chunks for one document: "
    f"{max(counts) if counts else 0})"
)


def propagate_qrels(
    qrels: dict[str, dict[str, int]],
) -> dict[str, dict[str, int]]:
    """Give every chunk of a relevant document that document's relevance."""

    qrels_chunks: dict[str, dict[str, int]] = {}
    for query_id, doc_scores in qrels.items():
        chunk_scores: dict[str, int] = {}
        for doc_id, relevance in doc_scores.items():
            for chunk in chunks_by_doc_id.get(doc_id, ()):
                chunk_scores[chunk.identifier] = relevance
        qrels_chunks[query_id] = chunk_scores
    return qrels_chunks


qrels_chunks: dict[str, dict[str, int]] = propagate_qrels(odex_task.qrels)

missing = sum(1 for scores in qrels_chunks.values() if not scores)
print(f"Queries left with zero relevant chunks after propagation: {missing}")
```

### Verify: every chunk of a gold document is actually covered

Check that the propagation above did what it claims. For every query, every chunk belonging to that query's gold document should now carry the document's relevance in `qrels_chunks`, whatever rank that chunk comes back at.

```python
mismatches = []

for query_id, doc_scores in odex_task.qrels.items():
    for doc_id, relevance in doc_scores.items():
        expected_chunk_ids = {
            chunk.identifier for chunk in chunks_by_doc_id.get(doc_id, ())
        }
        actual_scores = qrels_chunks[query_id]
        uncovered = {
            chunk_id
            for chunk_id in expected_chunk_ids
            if actual_scores.get(chunk_id) != relevance
        }
        if uncovered:
            mismatches.append((query_id, doc_id, uncovered))

if mismatches:
    raise AssertionError(
        f"{len(mismatches)} document(s) have chunks missing from "
        f"qrels_chunks after propagation -- first: {mismatches[0]}"
    )

print(
    "Every chunk of every gold document is present in qrels_chunks "
    "with the document's relevance."
)
```

## 7. Build a whole-document baseline (no splitting)

Build one unsplit chunk per corpus document -- the same granularity CodeRAG-Bench's own `qrels` already use, so no propagation is needed. Unlike real chunking, this can't split a long document into smaller pieces, so the outliers up to ~386,000 characters would otherwise be embedded as one enormous sequence and exceed the embedding model's memory; `WHOLE_DOCUMENT_MAX_CHARS` caps just this baseline's text, affecting about 1% of documents. Chunks are also sorted by length before embedding, so section 8's batching doesn't pad dozens of short documents up to one long outlier's size.

```python
WHOLE_DOCUMENT_MAX_CHARS = 20_000


def build_whole_document_chunks(task: BenchmarkTask) -> list[Chunk]:
    """Build one whole, unsplit chunk per corpus document.

    Real chunking (section 5) splits long documents into paragraph-sized
    pieces on its own, so it never sees a single oversized sequence. This
    baseline can't split anything -- it's one chunk per document by
    definition -- so the small number of outlier documents (up to ~386,000
    characters) are capped here to avoid encoding a sequence far beyond
    what the embedding model can handle in one pass.
    """

    chunks = []
    for doc_id, doc in task.corpus.items():
        text = doc["text"][:WHOLE_DOCUMENT_MAX_CHARS]
        chunks.append(
            Chunk(
                identifier=doc_id,
                path=Path("odex") / f"{doc_id}.txt",
                text=text,
                start_line=1,
                end_line=text.count("\n") + 1,
                kind="whole_document",
                symbol=doc.get("title") or None,
                section=None,
            )
        )
    # Sorted by length, not corpus order: batching (section 8) pads every
    # sequence in a batch to its longest member, so leaving a ~5,000-token
    # outlier next to dozens of ~80-token documents wastes almost all of
    # that batch's compute on padding. Grouping similar lengths together
    # keeps batches close to their true cost instead.
    chunks.sort(key=lambda chunk: len(chunk.text))
    return chunks


whole_document_chunks = build_whole_document_chunks(odex_task)
truncated = sum(
    1 for doc in odex_task.corpus.values() if len(doc["text"]) > WHOLE_DOCUMENT_MAX_CHARS
)
print(
    f"odex: {len(whole_document_chunks)} whole-document chunks "
    f"({truncated} truncated to {WHOLE_DOCUMENT_MAX_CHARS} characters)"
)
```

## 8. Run retrieval

Define semantic, lexical, and hybrid retrieval as reusable functions, then build the real-chunking index once. No code channel: ODEX's corpus is prose documentation, not parseable code, so there's nothing for `code_search` to resolve.

Embedding is the slow part -- so the index is saved under the project's own `indexes/` folder (the same mechanism `RAGSession` itself uses) and reloaded next time if its fingerprint still matches, rebuilt otherwise.

```python
import time

from tqdm.auto import tqdm

from benchmarks.common import format_query
from qachat import (
    build_lexical_index,
    build_semantic_index,
    hybrid_search,
    indexes_are_current,
    lexical_search,
    load_indexes,
    prepare_persistent_indexes,
    save_indexes,
    semantic_matrix,
    semantic_search,
)
from qachat.retrievals.code_graph.code_graph import CodeGraph

EMBED_BATCH_SIZE = 32
SEARCH_COUNT = 100
# No code entities exist for ODEX's prose corpus, so the
# fingerprint/caching machinery below -- built for QAChat's real
# semantic+lexical+code indexes -- is given a permanently empty code graph.
empty_code_graph = CodeGraph(entities=(), relations=())


def run_semantic_retrieval(
    chunks: list[Chunk],
    semantic_index,
    *,
    desc: str,
    matrix=None,
    count: int = SEARCH_COUNT,
) -> dict[str, dict[str, float]]:
    """Rank every query's chunks with QAChat's semantic search."""

    count = min(count, len(chunks))
    search_kwargs = {"backend": "numpy", "matrix": matrix} if matrix is not None else {}
    results: dict[str, dict[str, float]] = {}
    query_items = tqdm(odex_task.queries.items(), desc=f"{desc}: semantic", unit="query", leave=False)
    for query_id, query_text in query_items:
        ranked = semantic_search(
            semantic_index, format_query(query_text), embedding_session, count=count, **search_kwargs
        )
        results[query_id] = {result.chunk.identifier: result.score for result in ranked}
    return results


def run_lexical_retrieval(
    chunks: list[Chunk],
    lexical_index,
    *,
    desc: str,
    count: int = SEARCH_COUNT,
) -> dict[str, dict[str, float]]:
    """Rank every query's chunks with QAChat's lexical search."""

    count = min(count, len(chunks))
    results: dict[str, dict[str, float]] = {}
    query_items = tqdm(odex_task.queries.items(), desc=f"{desc}: lexical", unit="query", leave=False)
    for query_id, query_text in query_items:
        ranked = lexical_search(lexical_index, query_text, count=count)
        results[query_id] = {result.chunk.identifier: result.score for result in ranked}
    return results


def run_hybrid_retrieval(
    chunks: list[Chunk],
    semantic_index,
    lexical_index,
    *,
    desc: str,
    matrix=None,
    count: int = SEARCH_COUNT,
) -> dict[str, dict[str, float]]:
    """Fuse semantic and lexical rankings."""

    count = min(count, len(chunks))
    search_kwargs = {"backend": "numpy", "matrix": matrix} if matrix is not None else {}
    results: dict[str, dict[str, float]] = {}
    query_items = tqdm(odex_task.queries.items(), desc=f"{desc}: hybrid", unit="query", leave=False)
    for query_id, query_text in query_items:
        semantic_ranked = semantic_search(
            semantic_index, format_query(query_text), embedding_session, count=count, **search_kwargs
        )
        lexical_ranked = lexical_search(lexical_index, query_text, count=count)
        fused = hybrid_search(semantic=semantic_ranked, lexical=lexical_ranked, count=count)
        results[query_id] = {result.chunk.identifier: result.score for result in fused}
    return results


index_path = project.indexes / "odex_retrieval.json"
cached = None
if index_path.exists():
    try:
        cached = load_indexes(index_path)
    except ValueError:
        cached = None
    if cached is not None and not indexes_are_current(cached, odex_chunks, empty_code_graph):
        cached = None

if cached is not None:
    odex_semantic_index = cached.semantic
    odex_lexical_index = cached.lexical
    print(f"odex: loaded cached index ({len(odex_chunks)} chunks) from {index_path.relative_to(root)}")
else:
    with tqdm(total=len(odex_chunks), desc="Embedding odex chunks") as progress_bar:

        def _on_embed_progress(current: int, total: int) -> None:
            progress_bar.n = current
            progress_bar.refresh()

        odex_semantic_index = build_semantic_index(
            odex_chunks, embedding_session, batch_size=EMBED_BATCH_SIZE, progress_callback=_on_embed_progress
        )

    _lexical_start = time.monotonic()
    odex_lexical_index = build_lexical_index(odex_chunks)
    print(f"odex: lexical index built in {time.monotonic() - _lexical_start:.1f}s.")

    save_indexes(
        prepare_persistent_indexes(odex_semantic_index, odex_lexical_index, empty_code_graph),
        index_path,
    )
    print(f"odex: indexed {len(odex_chunks)} chunks, saved to {index_path.relative_to(root)}")

odex_matrix = semantic_matrix(odex_semantic_index)
```

## 9. Score against the published range

Scores against `qrels_chunks`, not the original document-level `odex_task.qrels` -- a hit on any chunk of a gold document counts. Semantic and lexical are scored alongside hybrid; with only two real channels, this table is already the full picture.

```python
from benchmarks.common import evaluate_retrieval

semantic_results = run_semantic_retrieval(odex_chunks, odex_semantic_index, desc="odex", matrix=odex_matrix)
lexical_results = run_lexical_retrieval(odex_chunks, odex_lexical_index, desc="odex")
hybrid_results = run_hybrid_retrieval(
    odex_chunks, odex_semantic_index, odex_lexical_index, desc="odex", matrix=odex_matrix
)

odex_scores = {
    "semantic": evaluate_retrieval(qrels_chunks, semantic_results),
    "lexical": evaluate_retrieval(qrels_chunks, lexical_results),
    "hybrid": evaluate_retrieval(qrels_chunks, hybrid_results),
}

for channel, scores in odex_scores.items():
    print(
        f"{channel:>10}: NDCG@10={scores['ndcg']['NDCG@10']:.3f}  "
        f"MRR@10={scores['mrr']['MRR@10']:.3f}  Recall@10={scores['recall']['Recall@10']:.3f}"
    )
```

**What the numbers show.** `hybrid` actually scores below `semantic` alone -- fusion isn't helping here. `lexical` is weak on its own, and RRF still gives its rankings some weight in the fused result, pulling the combined ranking away from semantic's better ordering.

This is a property of ODEX specifically, not of hybrid fusion in general: natural-language intents rarely share exact wording with the documentation prose that answers them, so BM25-style lexical matching has little to work with here.

## 10. Compare: whole-document baseline vs. QAChat chunking

Scores retrieval over `whole_document` chunks and over QAChat's real chunking (`qachat_chunking`, `hybrid`).

```python
with tqdm(total=len(whole_document_chunks), desc="Embedding odex whole documents") as progress_bar:

    def _on_embed_progress(current: int, total: int) -> None:
        progress_bar.n = current
        progress_bar.refresh()

    whole_document_semantic_index = build_semantic_index(
        whole_document_chunks, embedding_session, batch_size=EMBED_BATCH_SIZE, progress_callback=_on_embed_progress
    )
_lexical_start = time.monotonic()
whole_document_lexical_index = build_lexical_index(whole_document_chunks)
print(f"odex whole-document: lexical index built in {time.monotonic() - _lexical_start:.1f}s.")
whole_document_matrix = semantic_matrix(whole_document_semantic_index)

whole_document_results = run_hybrid_retrieval(
    whole_document_chunks,
    whole_document_semantic_index,
    whole_document_lexical_index,
    desc="odex whole-document",
    matrix=whole_document_matrix,
)

comparison_scores = {
    "whole_document": evaluate_retrieval(odex_task.qrels, whole_document_results),
    "qachat_chunking": odex_scores["hybrid"],
}

HEADLINE_MEASURES = (
    ("mrr", "MRR@10"),
    ("ndcg", "NDCG@10"),
    ("recall", "Recall@10"),
    ("recall", "Recall@100"),
)

header = ["mode", *(label for _, label in HEADLINE_MEASURES)]
print(" | ".join(f"{h:>14}" for h in header))
for mode in ("whole_document", "qachat_chunking"):
    scores = comparison_scores[mode]
    values = [f"{scores[group][label]:.3f}" for group, label in HEADLINE_MEASURES]
    print(" | ".join(f"{cell:>14}" for cell in [mode, *values]))
```

**Reading the two numbers.** `whole_document` scores nearly 3x higher than `qachat_chunking`. Splitting a document into several chunks spreads its relevance across more retrievable units, and a natural-language ODEX query doesn't always overlap well with the specific paragraph a chunk isolates, so more chunks per query here mostly means more ways to miss.

`whole_document`'s metrics hold up well against the published range despite QAChat's default embedding model being 0.6B parameters, running locally in 4-bit, next to GIST-large's 1024 dimensions, SFR-Mistral's 4096, or Voyage-code's proprietary embeddings behind the top scores.
