# Check the QAChat setup

This tutorial asks MLX or PyTorch to add two numbers. It just proves that the
selected calculation engine runs; it does not load a model.

The following code cells find QAChat and check its `.venv` environment.
Missing packages are repaired with `uv`.

If asked, select the printed `.venv` kernel and rerun the cell.

## Choose an engine

Use `auto`, `mlx`, or `torch`. On Apple Silicon, `auto` chooses MLX.

```python
BACKEND = "auto"
```

## Run the check

The selected engine calculates the result. Ordinary Python separately calculates `expected`. The assertion checks that both answers agree.

```python
A = 2
B = 3

expected = A + B
```

```python
from setup.check_setup import check_setup

report = check_setup(BACKEND, a=A, b=B)

report
```

```python
import math

assert math.isclose(report["result"], expected)
```

```python
print(f"Checked: {A} + {B} = {report['result']:g}")
```
