# Check the QAChat setup

This tutorial asks MLX or PyTorch to add two numbers. It proves that the selected calculation engine runs; it does not load a model.

On a new computer, first run the setup command in the README. Select QAChat's `.venv` kernel, then run one cell at a time with Shift+Enter.

```python
from pathlib import Path

root = Path.cwd()
if not (root / "qachat").is_dir():
    root = root.parent

print(f"Root directory: {root}")
```

```python
import sys

if str(root) not in sys.path:
    sys.path.insert(0, str(root))


print(f"Python environment: {sys.executable}")
```

```python
import platform

print(f"Computer system: {platform.system()}")
print(f"Computer machine: {platform.machine()}")
```

## Choose an engine and two numbers

Use `auto`, `mlx`, or `torch`. On Apple Silicon, `auto` chooses MLX. Change the two numbers to see that the printed answer is not fixed text.

```python
BACKEND = "auto"
A = 2
B = 3
```

## Run the real check

The selected engine calculates the result. Ordinary Python separately calculates `expected`. The assertion checks that both answers agree.

```python
import math
from setup.check_setup import check_setup

report = check_setup(BACKEND, a=A, b=B)

expected = A + B

assert math.isclose(
    report["result"],
    expected,
    rel_tol=1e-6,
    abs_tol=1e-6,
)
print(f"Checked: {A} + {B} = {report['result']:g}")

report
```
