# Tutorial 10 — Check the QAChat setup

This first tutorial verifies that the QAChat development environment is usable
before any model is loaded. It locates the repository, checks the `.venv`
environment, selects a **calculation backend**, and asks that backend to
perform a small numerical operation.

By the end of the tutorial, you will have verified three things:

- the notebook is running against the intended QAChat repository;
- the project environment and dependencies are available;
- MLX or PyTorch can execute a basic calculation on this computer.

No language model is loaded yet. The next tutorial introduces the shared
Hugging Face model cache.


## 1. Prepare QAChat

Each tutorial is independently runnable. These cells locate the repository,
make the local package importable, and call `prepare_notebook()` to verify the
project `.venv` environment. If Jupyter asks you to select the printed `.venv`
kernel, select it and rerun the cell.


```python
from pathlib import Path

root = Path.cwd()
if not (root / "qachat").is_dir():
    root = root.parent

print(f"QAChat root: {root}")
```

```python
import sys

if str(root) not in sys.path:
    sys.path.insert(0, str(root))
```

```python
from setup.notebook import prepare_notebook

root = prepare_notebook(root)
```

```python
import platform

print(f"Computer system: {platform.system()}")
print(f"Computer machine: {platform.machine()}")
```

## 2. Choose a backend

QAChat supports `auto`, `mlx`, and `torch`. `auto` resolves to MLX on supported
Apple Silicon systems and otherwise to PyTorch. Keeping the backend choice
explicit here makes later model selection easier to understand.


```python
BACKEND = "auto"
```

## 3. Run the backend check

The selected backend calculates a small result. Ordinary Python separately
computes `expected`, and the assertion verifies that both values agree. This is
a minimal execution test: success here confirms the backend is usable, not that
a language model has been downloaded or loaded.


```python
A = 2
B = 3

expected = A + B
```

```python
from setup.check_setup import check_setup

report = check_setup(BACKEND, a=A, b=B)

report
```

```python
import math

assert math.isclose(report["result"], expected)

print(f"Checked: {A} + {B} = {report['result']:g}")
```

## 4. Interpret the result

If the assertion passes and the checked result is printed, the QAChat project
environment and selected calculation backend are working. Model availability is
a separate concern and is handled through the Hugging Face cache in 
[Tutorial 20](./20_huggingface_models.ipynb).
