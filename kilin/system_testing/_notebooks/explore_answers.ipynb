# KíLín · Explore and compare answers

**Read-only · No replay · No model loading**

KíLín generates this `_answers.ipynb` for one reference workspace. The
batch pre-executes it from the same comparison snapshot as the sibling
comparison notebook. Open the generated notebook in `workspaces/` to
explore the saved outputs without running cells. Manually re-executing the
notebook reads the saved comparison JSON, not the historical questions.

See [README.md](../README.md) for details.

```python
from pathlib import Path
from IPython.display import HTML, display

from kilin.system_testing.batch import load_snapshot
from kilin.system_testing.analysis import (
    overview, event_inventory_html, question_explorer, operation_explorer,
)

REFERENCE = Path("__ORIGINAL_WORKSPACE__")
DESTINATION = REFERENCE.with_name(
    REFERENCE.name.removesuffix("_original") + "_replay"
)
data, source, digest = load_snapshot(REFERENCE)
print("Experiment:", REFERENCE.name)
print("Snapshot:", source)
print("Snapshot SHA-256:", digest)
```

## 1. Question-level overview

Check the recording validity and aligned question coverage before reading
individual pairs. If alignment is unavailable, independently saved answers are shown
when available, with invalid recordings explicitly marked as unverified.

```python
display(HTML(overview(data)))
```

## 2. Side-by-side answers and execution details

Expand a question to see the complete saved original and replay responses,
exact text, literal line diff, routing, retrieval, evidence, citations,
rankings, settings, timings and model calls.

Set `QUESTION` to a one-based matched-question index to focus on a case,
or `None` to display all. `DIFFERENCES_ONLY` filters for recorded changes.
`INCLUDE_MODEL_TEXT=True` reads exact local prompt/response text only when
its event type, sequence and SHA-256 match the report. Enable it selectively:
full prompts and answers may contain private information.

```python
QUESTION = None
DIFFERENCES_ONLY = False
INCLUDE_MODEL_TEXT = False

display(HTML(question_explorer(
    data, question=QUESTION, differences_only=DIFFERENCES_ONLY,
    original=REFERENCE, replayed=DESTINATION,
    include_model_text=INCLUDE_MODEL_TEXT,
)))
```

## 3. Event inventory and diagnostic differences

Open the full inventory or enable local trace details for event types
whose counts differ. The counts describe recorded diagnostics, not
necessarily different user operations. Local details are not hash-verified.

```python
INCLUDE_EVENT_DETAILS = False

display(HTML(event_inventory_html(
    data, original=REFERENCE, replayed=DESTINATION,
    include_details=INCLUDE_EVENT_DETAILS,
)))
```
